```python
from collections import Counter, defaultdict

words = "the cat sat on the mat the end".split()   # split the sentence into a list of words
counts = Counter(words)                 # counts how many times each word appears
print(counts.most_common(2))            # [('the', 3), ('cat', 1)]

groups = defaultdict(list)              # a dict that creates an empty list for any new key
for w in words:
    groups[len(w)].append(w)            # group words by their length; no KeyError on new keys
print(dict(groups))                     # {3: ['the', 'cat', 'sat', 'the', 'mat', 'the', 'end'], 2: ['on']}
```

```python
tags = ["ml", "ai", "ml", "nlp", "ai"]
unique = set(tags)             # {'ml', 'ai', 'nlp'} → duplicates removed (order not guaranteed)

a = {"python", "sql", "java"}
b = {"python", "pytorch"}
print(a & b)    # {'python'} → intersection: items in both
print(a | b)    # union: items in either
print(a - b)    # {'sql', 'java'} → difference: in a but not in b

print("sql" in a)   # True. Checking membership is O(1) for a set but O(n) for a list
```

```python
reviews = [
    ("r1", "Great room, great breakfast", 5),
    ("r2", "Room was dirty. Breakfast cold", 2),
    ("r3", "Great location, friendly staff", 4),
    ("r4", "Staff were rude, room was small", 2),
    ("r5", "Breakfast was great, staff friendly", 5),
]
```

```python
from collections import Counter, defaultdict

review_rating = {}
total_rating = 0
words_list = []
rating_dict = defaultdict(list)
positive_set = set()
negative_set = set()

for i in reviews:
    x,y,z = i
    print(x,"->",z,"★")

    review_rating[x] = z

    total_rating += z

    a = y.lower()
    a = a.replace(",","")
    a = a.replace(".","")
    b = a.split()
    words_list.extend(b)

    rating_dict[z].append(x)

    if z>=4:
        positive_set.update(b)
    elif z<=2:
        negative_set.update(b)

print(review_rating)

print(total_rating/5)

counts = Counter(words_list)
print(counts.most_common(3))

print(dict(rating_dict))

print(positive_set & negative_set)
```
