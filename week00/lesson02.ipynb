```python
# A lambda is a tiny function with no name, written in one line. Its most common use is telling sorted what to sort by

results = [("bert", 0.91), ("gpt", 0.95), ("lstm", 0.84)]

best = sorted(results, key=lambda r: r[1], reverse=True)
# key=lambda r: r[1] → "for each item r, sort by its second element (the score)"
# reverse=True → highest first
print(best[0])    # ('gpt', 0.95)

print(min(results, key=lambda r: r[1]))   # ('lstm', 0.84); max and min accept key= too
```

```python
reviews = [
    ("r1", "Great room, great breakfast", 5),
    ("r2", "Room was dirty. Breakfast cold", 2),
    ("r3", "Great location, friendly staff", 4),
    ("r4", "Staff were rude, room was small", 2),
    ("r5", "Breakfast was great, staff friendly", 5),
]
```

```python
def clean_text(text: str) -> list[str]:
    """ Return the cleaned text as a list of words

    Args:
        text: String of words with punctuation marks

    Returns:
        A list of words in lowercase
    """

    cleaned_text = text.replace(",","")
    cleaned_text = cleaned_text.replace(".","")
    cleaned_text_lower = cleaned_text.lower()
    cleaned_list = cleaned_text_lower.split()

    return cleaned_list

print(clean_text(reviews[0][1]))  #['great', 'room', 'great', 'breakfast']

def average_rating(reviews:list[tuple[str, str, int]], *, min_rating:int = 1) -> float:
    """ Return the average rating from all the reviews

    Args:
        reviews: List of tuples containing ID, review and ratings

    Returns:
        The average rating counting only those more than minimum of 1
    """
    total_rating = 0
    count_ratings = 0
    for review in reviews:
        id,review_text,rating = review
        if rating>=min_rating:
            total_rating+=rating
            count_ratings+=1
    
    if count_ratings == 0:
        return 0.0
    
    average = total_rating/count_ratings

    return average

print(average_rating(reviews)) #3.6

def top_reviews(reviews:list[tuple[str, str, int]], n=2) -> list[tuple[str, str, int]]:
    """ Return the top N reviews from all the reviews

    Args:
        reviews: List of tuples containing ID, review and ratings
        n: The number of top reviews needed. DFefault value is 2

    Returns:
        The list of top N reviews
    """
    top_ratings = sorted(reviews, key=lambda x:(-x[2],x[0]))
    return top_ratings[0:n]

print(top_reviews(reviews,3)) #[('r1', 'Great room, great breakfast', 5), ('r5', 'Breakfast was great, staff friendly', 5), ('r3', 'Great location, friendly staff', 4)]

def summarise(*review_ids: str, **options: bool) -> None:
    """Print the given review IDs, plus each review's text if verbose=True.

    Args:
        *review_ids: any number of review IDs, e.g. "r1", "r3"
        **options: optional flags; supports verbose=True

    Returns:
        None. Prints to the screen.
    """
    texts = {review_id: text for review_id, text, rating in reviews}

    for review_id in review_ids:                                       # review_ids is a tuple like ("r1", "r3")
        if options.get("verbose", False):                              # False if verbose wasn't passed
            print(f"{review_id}: {texts.get(review_id, 'not found')}") # .get handles an unknown ID safely
        else:
            print(review_id)

summarise("r1", "r3")
summarise("r1", "r3", verbose=True)
# r1
# r3
# r1: Great room, great breakfast
# r3: Great location, friendly staff

def collect(word:str, seen=None) -> set[str]:
    if seen is None:
        seen = set()
    seen.add(word)
    return seen

#seen is now None by default so each function call creates it's ownn set instead of reusing exisiting default which is created only once.
```
