```python
words = ["Great", "room", "a", "Breakfast", "ok"]

# The long way
lower = []
for w in words:
    lower.append(w.lower())

# List comprehension: [what to produce  for each item in source]
lower = [word.lower() for word in words]
# ['great', 'room', 'a', 'breakfast', 'ok']

# With a filter: [produce  for item in source  if condition]
long_words = [word.lower() for word in words if len(word) > 2]
# ['great', 'room', 'breakfast']

# Dict comprehension: {key: value for ...}
lengths = {word: len(word) for word in words}
# {'Great': 5, 'room': 4, 'a': 1, 'Breakfast': 9, 'ok': 2}

# Set comprehension: {value for ...}, which removes duplicates automatically
first_letters = {word[0].lower() for word in words}
# {'g', 'r', 'a', 'b', 'o'}
```

```python
def read_batches(items, batch_size):
    """Yield the items in chunks of batch_size."""
    for start in range(0, len(items), batch_size):  # 0, 2, 4, ...
        yield items[start:start + batch_size]       # yield: hand back ONE batch, then pause here

batches = read_batches([1, 2, 3, 4, 5], 2)
print(batches)          # <generator object ...> → nothing has run yet!
print(next(batches))    # [1, 2] → runs until the first yield, then pauses
for batch in batches:   # carries on from where it paused
    print(batch)        # [3, 4] then [5]
```

```python
def count_up_to(limit):
    print("starting")
    number = 1
    while number <= limit:
        print(f"about to yield {number}")
        yield number              # hand back number, then PAUSE here
        print(f"resumed after {number}")
        number += 1
    print("finished")

gen = count_up_to(3)
```

```python
print(gen)         # <generator object count_up_to ...>
                   # nothing printed! Calling the function only CREATES the generator; no code has run yet
```

```python
print(next(gen))   # "starting" → "about to yield 1" → 1
                   # ran from the top until the first yield, then paused
```

```python
print(next(gen))   # "resumed after 1" → "about to yield 2" → 2
                   # carried on right after the yield, went round the loop once, paused again
```

```python
print(next(gen))   # "resumed after 2" → "about to yield 3" → 3
```

```python
print(next(gen))   # "resumed after 3" → "finished" → StopIteration error
                   # the function ended without another yield, so there's nothing left to give
```

```python
for number in count_up_to(3):
    print(number)       # 1, 2, 3, then the loop ends cleanly
```

```python
def make_batches(items, batch_size):
    """Yield items in chunks of batch_size."""
    for start in range(0, len(items), batch_size):   # start = 0, 2, 4, ...
        yield items[start:start + batch_size]        # one slice at a time

for batch in make_batches(["a", "b", "c", "d", "e"], 2):
    print(batch)
# ['a', 'b']
# ['c', 'd']
# ['e']     ← the last batch is smaller; the slice just stops at the end of the list
```

```python
def stream_reviews(reviews):
    for review_id,review_text,rating in reviews:
        formatted_review = review_id+" | "+(rating*"★")+" | "+review_text
        yield formatted_review

for review in stream_reviews(reviews):
    print(review)

top_review_count = sum(1 for review_id,review_text,rating in reviews if rating>=4)
print(top_review_count)
```

```python
y_true = [1, 0, 1, 1]      # correct labels
y_pred = [1, 0, 0, 1]      # what the model predicted

for i in range(len(y_true)):
    print(y_true[i], y_pred[i])   # works, but the indexing is clumsy

for true, pred in zip(y_true, y_pred):
    print(true, pred)
# 1 1
# 0 0
# 1 0   ← the model got this one wrong
# 1 1

correct = sum(true == pred for true, pred in zip(y_true, y_pred))
# true == pred gives True or False for each pair → True counts as 1 → correct = 3
accuracy = correct / len(y_true)    # 0.75
```

```python
def log_epoch(epoch: int, loss: float, path: str = "train.log") -> None:
    """Append one line per epoch to a log file."""
    with open(path, "a", encoding="utf-8") as file:        # "a" = keep the old lines
        file.write(f"epoch={epoch:02d} loss={loss:.4f}\n")

for epoch, loss in enumerate([0.9, 0.6, 0.4], start=1):
    log_epoch(epoch, loss)
# train.log now contains:
# epoch=01 loss=0.9000
# epoch=02 loss=0.6000
# epoch=03 loss=0.4000
```

```python
import json

ratings = [rating for review_id, review_text, rating in reviews]

summary = {
    "count": len(reviews),                                          # 5
    "average_rating": round(sum(ratings) / len(ratings), 2),        # 3.6
    "top_ids": [review_id for review_id, review_text, rating in reviews if rating == max(ratings)],  # ['r1', 'r5']
    "positive_ratio": sum(rating >= 4 for rating in ratings) / len(ratings),  # 3/5 = 0.6
}

with open("summary.json", "w", encoding="utf-8") as file:
    json.dump(summary, file, indent=2)

with open("summary.json", encoding="utf-8") as file:
    loaded = json.load(file)

print(loaded["average_rating"])   # 3.6
```
