```python
words = ["Great", "room", "a", "Breakfast", "ok"]

# The long way
lower = []
for w in words:
    lower.append(w.lower())

# List comprehension: [what to produce  for each item in source]
lower = [word.lower() for word in words]
# ['great', 'room', 'a', 'breakfast', 'ok']

# With a filter: [produce  for item in source  if condition]
long_words = [word.lower() for word in words if len(word) > 2]
# ['great', 'room', 'breakfast']

# Dict comprehension: {key: value for ...}
lengths = {word: len(word) for word in words}
# {'Great': 5, 'room': 4, 'a': 1, 'Breakfast': 9, 'ok': 2}

# Set comprehension: {value for ...}, which removes duplicates automatically
first_letters = {word[0].lower() for word in words}
# {'g', 'r', 'a', 'b', 'o'}
```

```python
def read_batches(items, batch_size):
    """Yield the items in chunks of batch_size."""
    for start in range(0, len(items), batch_size):  # 0, 2, 4, ...
        yield items[start:start + batch_size]       # yield: hand back ONE batch, then pause here

batches = read_batches([1, 2, 3, 4, 5], 2)
print(batches)          # <generator object ...> → nothing has run yet!
print(next(batches))    # [1, 2] → runs until the first yield, then pauses
for batch in batches:   # carries on from where it paused
    print(batch)        # [3, 4] then [5]
```

```python
import sys
squares_list = [n * n for n in range(1_000_000)]   # builds 1M numbers in memory now
squares_gen  = (n * n for n in range(1_000_000))   # builds nothing yet

print(sys.getsizeof(squares_list))   # ~8,000,000 bytes
print(sys.getsizeof(squares_gen))    # ~200 bytes

print(sum(n * n for n in range(1_000_000)))   # inside a function call you can drop the extra brackets
```
